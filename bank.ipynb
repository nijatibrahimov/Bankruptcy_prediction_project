```python
def remove_high_vif(X, threshold=10):

    X = pd.DataFrame(X)
    while True:
        vif = pd.DataFrame()
        vif["feature"] = X.columns
        vif["VIF"] = [
            variance_inflation_factor(X.values, i)
            for i in range(X.shape[1])
        ]
        
        max_vif = vif["VIF"].max()
        
        if max_vif > threshold:
            drop_feature = vif.sort_values("VIF", ascending=False)["feature"].iloc[0]
            print(f"Silinir: {drop_feature} (VIF={max_vif})")
            X = X.drop(columns=[drop_feature])
        else:
            break
            
    return X

X_final = remove_high_vif(x)
```

```python
X_df = pd.DataFrame(X_final)

vif_data = pd.DataFrame()
vif_data["feature"] = X_df.columns
vif_data["VIF"] = [
    variance_inflation_factor(X_df.values, i)
    for i in range(X_df.shape[1])
]

print(vif_data.sort_values(by="VIF", ascending=False))
```

```python
import pandas as pd
import numpy as np

# Tutaq ki, X_final feature-lər, y isə target
# Covariance və correlation hesablamaq üçün:

def variable_influence_report(X, y):
    report = []
    for col in X.columns:
        cov = np.cov(X[col], y)[0, 1]  # covariance
        corr = np.corrcoef(X[col], y)[0, 1]  # correlation
        report.append({
            'variable': col,
            'covariance': cov,
            'correlation': corr,
            'interpretation': 'strong' if abs(corr) > 0.5 else ('moderate' if abs(corr) > 0.1 else 'weak')
        })
    return pd.DataFrame(report).sort_values(by='correlation', key=abs, ascending=False)

# Nümunə çağırış
influence_df = variable_influence_report(X_final, y)
print(influence_df)
```

```python
# weak correlation-ları çıxarırıq
strong_moderate_vars = influence_df[influence_df['interpretation'] != 'weak']['variable']

# X_scaled üzərində weak-ları silirik
X_scaled_filtered = X_final[strong_moderate_vars]

# Nəticəni yoxlayaq
print("Filtered X_scaled shape:", X_scaled_filtered.shape)
print("Keçmiş columns:", X_final.columns.tolist())
print("Qalan columns:", X_scaled_filtered.columns.tolist())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, classification_report

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X_scaled_filtered, y, test_size=0.2,stratify=y, random_state=42)

```

```python
# X_train: DataFrame (14 feature)
# y_train: Series (target column, məsələn 'Bankrupt')

yy_train = y_train.reset_index(drop=True)
XX_train = X_train.reset_index(drop=True)

train_df = XX_train.copy()
train_df['Bankruptcy'] = yy_train

train_df.to_csv("bankruptcy_model_data.csv", index=False)

print("Merged DataFrame hazır, shape:", train_df.shape)
```

```python
data = pd.read_csv("bankruptcy_model_data.csv")
X_train = data.drop('Bankruptcy', axis=1)
y_train = data['Bankruptcy']
```

```python
import numpy as np

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
    precision_recall_curve
)

# 1. Pipeline
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('logistic_model', LogisticRegression(max_iter=3000))
])

# 2. Hyperparameter grid
param_grid = {
    'logistic_model__C': [ 0.1, 1, 10, 100],
    'logistic_model__solver': ['liblinear', 'lbfgs'],
    'logistic_model__class_weight': [None, 'balanced']
}

# 3. Cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# 4. Grid Search
grid = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    scoring='f1',
    cv=cv,
    n_jobs=-1,
    verbose=1
)

grid.fit(X_train, y_train)

print("Best Parameters:", grid.best_params_)
print("Best CV F1-score:", grid.best_score_)

best_model = grid.best_estimator_

val_proba = best_model.predict_proba(X_test)[:, 1]

precision, recall, thresholds = precision_recall_curve(
    y_test, val_proba
)

min_precision = 0.40

valid_indices = np.where(precision[:-1] >= min_precision)[0]

if len(valid_indices) > 0:
    f1_values = (
        2 * precision[valid_indices] * recall[valid_indices]
        / (precision[valid_indices] + recall[valid_indices] + 1e-10)
    )

    best_idx = valid_indices[np.argmax(f1_values)]
    best_threshold = thresholds[best_idx]

    print("Selected threshold:", best_threshold)
    print("Validation precision:", precision[best_idx])
    print("Validation recall:", recall[best_idx])

else:
    best_threshold = 0.5
    print("Precision hədəfinə çatan threshold tapılmadı.")

test_proba = best_model.predict_proba(X_test)[:, 1]
y_pred = (test_proba >= best_threshold).astype(int)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

print("Class 1 Precision:", precision_score(y_test, y_pred, zero_division=0))
print("Class 1 Recall:", recall_score(y_test, y_pred, zero_division=0))
print("Class 1 F1:", f1_score(y_test, y_pred, zero_division=0))
```
